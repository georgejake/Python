```python
'''
Reference : https://www.youtube.com/watch?v=3dEPY3HiPtI (youtube video)
https://www.geeksforgeeks.org/multithreading-python-set-1/
'''
import threading
import time

def havingbreakfast():
    time.sleep(4)
    print("I am having breakfast")
def havingcoffee():
    time.sleep(4)
    print("I am having coffee")

def study():
    time.sleep(4)
    print("I am study")

x = threading.Thread(target=havingbreakfast,args=())
x.start()

y = threading.Thread(target=havingcoffee,args=())
y.start()

z = threading.Thread(target=study,args=())
z.start()

#Synchronizing thread with Main
#x.join()
#y.join()
#z.join()

print(threading.active_count())
print(threading.enumerate())
```

```python
'''
Example from Dhaval's code
'''
import time
import threading
from threading import Thread

def sleepMe(i):
    print("Thread %i will sleep." % i)
    time.sleep(5)
    print("Thread %i is awake" % i)

for i in range(10):
    th = Thread(target=sleepMe, args=(i, ))
    th.start()
    print("Current Threads: %i." % threading.active_count())
```
