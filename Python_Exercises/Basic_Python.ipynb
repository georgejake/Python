```python
#Dictionaries in python ,this is something of type key/value pair
d = {"george": "1234", "ditta": "1234", "diya": 123, "dhwani": 345}
print(d["george"])
print(d)
print("whole dictionary")
del d["george"]
print(d)
d["george"] = "1234"
#Checking whether a key named george is there in dictionary
print("george" in d)
for key in d:
    print(key)
    print(d[key])
for k, v in d.items():
    print("Key is:", k, "Value is:", v)
```

```python
#Functions Parameters can be passed by position and by name. There is also provision to provide optional parameter
def total_cost(expenses):
    total = 0
    for item in expenses:
        total = total+item
    return total

joe_expenses = [1,2,3]
print(total_cost(joe_expenses))

def sum(a,b=1):
    return a+b

print(sum(a=2,b=4))
```

```python
"""
In PYTHON ,Dictionary closely resembles json. 
Initialize a JSON, Dump JSON to string,Dump JSON to a file, Read JSON document and parse
Why relative path using dot natoation is not working
"""
import json
books = {}
books["tom"]={"name":"george","age":"43"}
books["ditta"]={"name":"ditta","age":"35"}
print(books)
print(json.dumps(books))
fromjson = json.loads('{"name":"ditta","age":"35"}')
print(fromjson)
for k,v in fromjson.items():
    print("key is" + k)
    print("value is" + v)

"""
Printing nested keys' values
"""
print(books["tom"]["age"])
print(books["tom"])
for keys in books:
    print(books[keys]["name"])
```
