```python
'''
https://www.geeksforgeeks.org/generators-in-python/, generators has many advantages in python. Most important is memory optimization
'''
def generator_of_squares(limit):
    for x in range (1,limit):
        y = x**2
        yield y
        

squares_seqs = generator_of_squares(5)

for square in squares_seqs:
    print(square)
    
```
