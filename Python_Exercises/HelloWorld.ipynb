```python
spendings = [1346.0, 987.50, 1734.40, 2567.0, 3271.45, 2500.0, 2130.0, 2510.30, 2987.34, 3120.50, 4069.78, 1000.0]
low_spendings = 0
normal_spendings = 0
high_spendings = 0
for spending in spendings:
    if spending < 1000:
        low_spendings+=1
    elif (spending >= 1000 and spending <=2500):
        normal_spendings+=1
    else:
        high_spendings+=1

print('Numbers of months with low spendings: ',low_spendings,', normal spendings: ', normal_spendings,', high spendings: ',high_spendings,'.',sep='')
print('Numbers of months with low spendings: ' + str(low_spendings)+', normal spendings: '+str(normal_spendings)+', high spendings: '+str(high_spendings)+'.')
```

```python
#List Comprehensions

i = []
for count in range(1,100):
    i.append(count)
print(i)
#above is condensed
j = [count for count in range(1,100)]
print(j)

#custom logic
k = [count for count in range(1,100) if count%3 !=0]
print(k)
```

```python
#Introduction to Tuples, tuples are similar to list ;main difference is tuples are immutable where as list is mutable
my_tuple = 1,
print(my_tuple)
my_tuple_1 = (1,2,)
print(my_tuple_1)
#Tuple operations
print(len(my_tuple_1))
if 3 in my_tuple_1:
    print('You are in')
else:
    print('Not in')
for element in my_tuple_1:
    print(element)
my_tuple_2 = my_tuple_1 + ('teacher','student')
print(my_tuple_2)
```
