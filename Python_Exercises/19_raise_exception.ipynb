```python
'''
https://www.geeksforgeeks.org/define-custom-exceptions-in-python/ ; also understand that in custom exception class we always overrid __str__(self) 
function
'''
class Person:
    def __init__(self,age,name):
        self.age = age
        self.name = name
    def display_person(self):
        print(f"Name is {self.name} and age is {self.age}")
    def get_minor_age(self):
        if self.age > 18:
            return self.age
        else:
            # Raising excepton
            raise AdultException("You are not adult")
           
#Custom exception class inherited from Exception class
class AdultException(Exception):
    def __init__(self,errorMessage):
        super().__init__()
        self.errorMessage = errorMessage
    #__str__(self) ,str function is overridden in custom exception class to represent the exception object as a string message
    def __str__(self):
        return f"There is an exception {self.errorMessage}"

p1 = Person(10,"George")
p1.display_person()
#Catching exception
try:
    p1.get_minor_age()    
except  AdultException as e:
    print(f"Caught error {e}")
    

    
```

```python
#More simplified example from course
class AdultException(Exception):
    pass


class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def get_minor_age(self):
        if int(self.age) >= 18:
            raise AdultException
        else:
            return self.age

    def display(self):
        try:
            #Inside try block the get_minor_age is invoked which raises AdultException
            print(f"age -> {self.get_minor_age()}")
        except AdultException:
            print("Person is an adult")
        finally:
            print(f"name -> {self.name}")


# No exception
Person("Bhavin", 17).display()

# AdultException is raised
Person("Dhaval", 23).display()
```
