```python
'''
poem.txt contains famous poem 
"Road not taken" by poet Robert Frost. You have to read this file in your python program 
and find out words with maximum occurance.
#in below comment also backslash is replaced with double backslash
Programming languages, such as Python, treat a backslash (\\) as an escape character.
For instance, \\n represents a line feed, and \\t represents a tab.
When specifying a path, a forward slash (/) can be used in place of a backslash.

with open("C:/Users/George/Python_Notebooks/resources/poem.txt","r") as readFile:
    s=readFile.read()

'''

words_and_count = {}
with open("C:/Users/George/Python_Notebooks/resources/poem.txt","r") as readFile:    
    for line in readFile:
        #string.split(separator, maxsplit)
        for word in line.split():
            if word in words_and_count:
                words_and_count[word] += 1
            else:
                words_and_count[word] = 1

print(words_and_count)
max_count=0
max_count_word = []
for word,count in words_and_count.items():
    if count == max_count:
        max_count_word.append(word)
        max_count = count
    elif count > max_count:
        max_count_word.clear()
        max_count_word.append(word)
        max_count = count

print(f"Words which have maximum count is {max_count_word}")
print(f"Max count is {max_count}")

#Another logic to find max word_count

word_occurances = list(words_and_count.values())
max_count = max(word_occurances)
print("Max occurances of any word is:",max_count)

print("Words with max occurances are: ")
for word, count in words_and_count.items():
    if count==max_count:
        print(word)
```

```python
'''
stocks.csv contains stock price, earnings per share and book value. You are writing a stock market application that will process this file and
create a new file with financial metrics such as pe ratio and price to book ratio. These are calculated as,
pe ratio = price / earnings per share
price to book ratio = price / book value
'''
out_string = f"CompanyName,PE Ratio,PB Ratio\n"
with open("C:/Users/George/Python_Notebooks/resources/stocks.csv","r") as readFile:    
    line_count = 1    
    for line in readFile: 
        # skip first line as it is heading
        if line_count == 1:
            line_count += 1
            continue        
        #string.split(separator, maxsplit)
        #separator - Optional. Specifies the separator to use when splitting the string. By default any whitespace is a separator
        #maxsplit - Optional. Specifies how many splits to do. Default value is -1, which is "all occurrences"
        elem_pos = 1  
        company_name = ""
        price = 0
        eps = 0
        bookvalue = 0
        for elem in line.split(","):
            if elem_pos == 1:
                company_name = elem
            elif elem_pos == 2:
                price = float(elem)
            elif elem_pos == 3:
                eps = float(elem)
            elif elem_pos == 4:
                bookvalue = float(elem)
            elem_pos += 1
        pe_ratio = round(price / eps,2)
        pb_ratio = round(price / bookvalue,2)
        
        out_string = out_string + f"{company_name},{pe_ratio},{pb_ratio}\n"

with open("C:/Users/George/Python_Notebooks/resources/stocks_ratios.csv","w") as writeFile:
    writeFile.write(out_string)           
            
```
