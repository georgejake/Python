```python
'''
https://www.w3schools.com/python/python_inheritance.asp
'''

class Animal:    
    def __init__(self,habitatName):
        self.habitatName = ""
        print("Please override in child class")
        

class Dog(Animal):
    def __init__(self,habitatName):
        super().__init__("Generic")
        self.habitatName = habitatName        
    def displayHabitat(self):
        print(f"Dog's habitatname is {self.habitatName}")
    

dog1 = Dog("kennel")
dog1.displayHabitat()

           
           
```

```python
#More concepts from Dhaval's repo
class Animal:
    def __init__(self, habitat):
        self.habitat = habitat

    def print_habitat(self):
        print(self.habitat)

    def sound(self):
        print("Some Animal Sound")


class Dog(Animal):
    def __init__(self):
        super().__init__("Kennel")

    def sound(self):
        print("Woof woof!")


x = Dog()
x.print_habitat()
x.sound()
```
