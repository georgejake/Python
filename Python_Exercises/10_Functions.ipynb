```python
'''
Write a function called calculate_area that takes base and height as an input and returns and area of a triangle. Equation of an area of a triangle is,
area = (1/2)*base*height
'''
def calculate_area(base,height):
    return ((1/2) * base * height)

print(calculate_area(5,2))
```

```python
'''
Modify above function to take third parameter shape type. It can be either "triangle" or "rectangle". 
Based on shape type it will calculate area. Equation of rectangle's area is,rectangle area=length*width
'''
def calculate_area_4_shapes(base,height,shape):
    if shape == 'rectangle':
        return (base * height)
    else:
        return ((1/2) * base * height)

print(calculate_area_4_shapes(5,2,"triangle"))    
```
