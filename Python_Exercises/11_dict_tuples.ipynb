```python
'''
We have following information on countries and their population (population is in crores)
China-143/India-146/USA-32/Pakistan-31
'''
#Use above to create a dictionary of countries and population
country_population = { }
country_population["China"] = {"population":143}
country_population["India"] = {"population":146}
country_population["USA"] = {"population":32}
country_population["Pakistan"] = {"population":31}

print(country_population)
'''
Write a program that asks user for three type of inputs,
print: if user enter print then it should print all countries with their population in this format
add: if user input add then it should further ask for a country name to add. If country already exist in our dataset then it should print that it exist and do nothing. If it doesn't then it asks for population and add that new country/population in our dictionary and print it
remove: when user inputs remove it should ask for a country to remove. If country exist in our dictionary then remove it and print new dictionary using format shown above in (a). Else print that country doesn't exist!
query: on this again ask user for which country he or she wants to query. When user inputs that country it will print population of that country.
'''
def print_countrypopulation():
    for key in country_population:
        print(key+"==>"+ str(country_population[key]["population"]))

userinput = input("Enter any of the choices - print/add/remove/query")

if userinput == "print":
    print_countrypopulation()
elif userinput == "add":
    cname = input("Enter country name")
    if cname in country_population:
        print("Country already exists")
    else:
        cpopulation = int(input("Population"))
        country_population[cname] = {"population":cpopulation}
        print_countrypopulation()
elif userinput == "remove":
    cname = input("Enter country name")
    if cname not in country_population:
        print("Country not exists")
    else:
        del country_population[cname]
        print_countrypopulation()
elif userinput == "query":
    cname = input("Enter country name")
    if cname in country_population:
        print("Population of country "+cname,str(country_population[cname]["population"]),sep="-")
    else:
        print("Country not found")
```

```python
'''
You are given following list of stocks and their prices in last 3 days,
info[600,630,620]/ril[1430,1490,1567]/mtl[234,180,160] 
print: When user enters print it should print following
info ==> [600, 630, 620] ==> avg:  616.67
ril ==> [1430, 1490, 1567] ==> avg:  1495.67
mtl ==> [234, 180, 160] ==> avg:  191.33
add: When user enters 'add', it asks for stock ticker and price.
If stock already exist in your list (like info, ril etc) then it will append the price to the list.
Otherwise it will create new entry in your dictionary.
For example entering 'tata' and 560 will add tata ==> [560] to the dictionary of stocks.
'''
price_of_stock = {
    "info":[600,630,620],
    "ril":[1430, 1490, 1567],
    "mtl":[234, 180, 160]    
}

def print_stockname_average():
    for k,v in price_of_stock.items():
        avg = round(sum(v)/len(v),2)
        print(f"{k} ==> {v} ==> avg: {avg}")

def add_stock_and_price():
    ticker_name = input("Enter ticker name")
    ticker_price = float(input("Enter ticker price"))
    if ticker_name in price_of_stock:
        price_of_stock[ticker_name] .append(ticker_price)
    else:
        price_of_stock[ticker_name] = [ticker_price]
    print_stockname_average()
    

def main():
    op=input("Enter operation (print, add or amend):")
    if op.lower() == 'print':
        print_stockname_average()
    elif op.lower() == 'add':
        add_stock_and_price()
    else:
        print("Unsupported operation:",op)

if __name__ == '__main__':
    main()
```

```python
'''
Write circle_calc() function that takes radius of a circle as an input from user and then it calculates and returns area,circumference and diameter.
You should get these values in your main program by calling circle_calc function and then print them
'''
import math

def circle_calculation(radius):
    area = round(math.pi * (radius**2),2)
    circumfrence = round(2 * math.pi * radius,2)
    diameter = round(2*radius,2)
    return (area,circumfrence,diameter)

if __name__ == "__main__":
    r = float(input("Enter radius of circle"))
    area,circumfrence,diameter = circle_calculation(r)
    print(f"area is {area},circumfrence is {circumfrence},diameter is {diameter}")


```
